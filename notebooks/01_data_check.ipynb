```python
from pathlib import Path
import imdlib as imd

here = Path.cwd()
ROOT = here.parent if here.name == "notebooks" else here   # find project root
RAW = ROOT / "data" / "raw" / "imd"
RAW.mkdir(parents=True, exist_ok=True)                    # create all missing levels
print("Saving to:", RAW)

for var in ["rain", "tmax"]:
    imd.get_data(var, 2020, 2020, fn_format="yearwise", file_dir=str(RAW))
```

```python
for f in sorted(RAW.rglob("*")):
    if f.is_file():
        print(f.relative_to(RAW), round(f.stat().st_size / 1e6, 1), "MB")
```

```python
rain = imd.open_data("rain", 2020, 2020, "yearwise", str(RAW)).get_xarray()
rain = rain.where(rain["rain"] != -999.)
rain
```

```python
tmax = imd.open_data("tmax", 2020, 2020, "yearwise", str(RAW)).get_xarray()
print(tmax)
print("Max value:", float(tmax["tmax"].max()))
```
